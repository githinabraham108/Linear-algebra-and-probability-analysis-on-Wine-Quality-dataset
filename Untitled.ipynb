```python
X=data[["pH","quality"]]
print("the feature matrix:\n ",X)
```

```python
cov_matrix=np.cov(X.T)
print("the covariance matrix of X:\n", cov_matrix)
```

#### TASK 4

```python
e_value,e_vector=np.linalg.eig(cov_matrix)
```

```python
x=data["quality"].mode().iloc[0]
print(f"most common wine quality is {x}")
```

```python
sns.displot(data=data["quality"],kind="hist")
plt.show()
```

```python
q_mean=data["quality"].mean()
q_median=data["quality"].median()
q_mode=x
q_std=data["quality"].std()
print(f"the mean of the data is {q_mean:.2f}")
print(f"the median of the data is {q_median:.2f}")
print(f"the mode of the data is {q_mode}")
print(f"the data has a std of {q_std:.2f}")
print(f"a std value of {q_std:.2f}, which is close to 1, means that the data is mostly concentrated around 1 std +/- from the mean value.")
```

```python
q_skew= round(data["quality"].skew(),2)
print(f"the data has a skewness of {q_skew}. Implies that the data is close to symmetric with slight skewness to right. ")
```

```python
q_kurt=round(data["quality"].kurt(),2)
print(f"the data has a  excess kurtic value of {q_kurt}, meaning that the distribution is not fully lepto nor fully normal in nature. Rather has moderate heavy tails with peaks.")
```
